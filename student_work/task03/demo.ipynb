```python
import sys 
from pathlib import Path
project_root = Path("../../").resolve()
sys.path.append(str(project_root))

from pprint import pprint
from student_work.task03.support import compute_support
from student_work.task03.rule import identify_task
```

```python
setting = {
    "adjacency":[
        [0,1,1,1],
        [1,0,1,1],
        [1,1,0,1],
        [1,1,1,0],
    ],
    "probabilities":[
        0.2,
        0.4,
        0.6,
        0.8
    ],
1,
    "assumptions":{
        "fixed_pretreatment_network": True,
        "fixed_pretreatment_core_sets": True,
        "known_assignment_law": True,
        "full_vector_randomization": True,
        "consistency": True,
        "finite_conditional_first_moments": True,
    },
}
print("节点数量: ",len(setting["probabilities"]))
print("treatment 概率：", setting["probabilities"])
print("核心合作伙伴：", setting["core_sets"])
```

```python
for i, core_set in enumerate(setting["core_sets"]):
    threshold = len(core_set) // 2 + 1
    print(
        f"节点 {i}: C_i={core_set}, "
        f"至少 {threshold} 名核心伙伴 treated 时 S_i=1"
    )
```

```python
support = compute_support(setting)

for node_id, arms in support.items():
    print(f"\n节点 {node_id}")
    for arm, probability in arms.items():
        print(f"  π_{node_id}{arm} = {probability:.4f}")
```

```python
result = identify_task(setting)
print("Identification status:", result["status"])
print("Functional:", result.get("functional"))
```

所有节点的四个arm概率都大于0，关键假设也已确认，因此：
mu_i(t,s) = E[Y_i^obs | A,X,T_i = t,S_i =s]
由观察总体分布唯一确定，所以结果为POINT
